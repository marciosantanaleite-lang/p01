```python
df_1 = con.execute("""
     SELECT * 
    FROM bronze_z0019
    WHERE nome_arquivo = 'z0019_2.csv'
""").fetchdf()
df_1.head(10)
```

```python
df_2 = con.execute("""
     SELECT * 
    FROM bronze_z0019
    WHERE nome_arquivo = 'z0019_1.csv'
""").fetchdf()
df_2.head(10)
```

```python
df_final = duckdb.sql("""
            SELECT *
            FROM df_1
            UNION ALL
            SELECT * FROM df_2
      """).fetchdf()
df_final.head(10)
```

```python
df = con.execute("""
            SELECT * FROM df_final
            ORDER BY data_ingestao
""").fetchdf()
df.head(10)
```

```python
df = con.execute("""
                SELECT * FROM (
                 SELECT *, ROW_NUMBER() OVER (PARTITION BY NATBR ORDER BY data_ingestao DESC) AS row,
                 FROM df_final
                WHERE data_ingestao >= '2025-06-16')
                WHERE row = 1
""").fetchdf()
df.head(30)

```

```python

df_final = df.rename(columns={'NATBR':'id', 
                              'MAKTX':'nm_produto', 
                              'WERKS':'nm_categoria', 
                              'MAINS':'id_fornecedor',
                               'LABST':'vl_preco' })

df_final = df_final.drop(columns=['nome_arquivo', 'data_ingestao', 'row'])

df_final.head(10)
```

```python
df_final.dtypes
```

```python
df2 = df_final.astype({'id': 'int64',
                       'nm_produto': 'string',
                          'nm_categoria': 'string',
                            'id_fornecedor': 'int64',
                            'vl_preco': 'float64'})
df2.dtypes
```

```python
con.execute("""
CREATE TABLE IF NOT EXISTS produtos
(
    id INTEGER,
    nm_produto STRING,
    nm_categoria STRING,
    id_fornecedor INTEGER,
    vl_preco FLOAT
)
            
""")
```
